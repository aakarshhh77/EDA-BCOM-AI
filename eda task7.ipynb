```python
categorical_cols = [
    "gender",
    "race/ethnicity",
    "parental level of education",
    "lunch",
    "test preparation course"
]

# Remove leading/trailing spaces
# Convert text to lowercase
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
```

```python
for col in categorical_cols:
    print("\n", col)
    print(df[col].unique())
```

```python
score_cols = [
    "math score",
    "reading score",
    "writing score"
]
```

```python
statistics = pd.DataFrame({

    "Mean": df[score_cols].mean(),

    "Median": df[score_cols].median(),

    "Standard Deviation": df[score_cols].std(),

    "Q1": df[score_cols].quantile(0.25),

    "Q2 (Median)": df[score_cols].quantile(0.50),

    "Q3": df[score_cols].quantile(0.75)

})
```

```python
statistics
```

```python
print(df[score_cols].describe())
```

```python
df["Total Marks"] = df[score_cols].sum()
```

```python
df["Percentage"] = (
    df["Total Marks"] / 300
) * 100
```

```python
print(
    df[
        score_cols +
        ["Total Marks", "Percentage"]
    ])
```

```python
performance = []

for percentage in df["Percentage"]:

    if percentage >= 90:
        performance.append("Excellent")

    elif percentage >= 75:
        performance.append("Very Good")

    elif percentage >= 60:
        performance.append("Good")

    elif percentage >= 50:
        performance.append("Average")

    else:
        performance.append("Needs Improvement")


df["Performance"] = performance

print(df[["Percentage", "Performance"]])
```

```python
for col in ["math score", "reading score", "writing score"]:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print("\nOutliers in", col)
    print(outliers[col])
```
